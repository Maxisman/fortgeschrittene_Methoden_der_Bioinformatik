```python
import networkx as nx
import random
import pandas as pd

G = nx.DiGraph()

for i in range(6):
    G.add_node(i, color = random.choice(["red","yellow","blue"]))
    
for i in range(6):
    for j in range(6):
        if i != j and random.random()< .3:
            G.add_edge(i,j)
            
```

```python
colornodedict = nx.get_node_attributes(G,"color")
cherrylist = []
for key1, value1 in colornodedict.items():
    for key2, value2 in colornodedict.items():
        if key1 < key2 and value1 != value2:
            cherrylist.append(frozenset({key1, key2}))
```

```python
from BICcherry import BICcherry
import networkx as nx
import matplotlib
matplotlib.use("TkAgg")   
import matplotlib.pyplot as plt


G = nx.DiGraph()
G.add_node('x', color='blue')
G.add_node('y', color='red')
G.add_node('z', color='red')
G.add_edge('y', 'x')
G.add_edge('z', 'x')
G.add_edge('x', 'z')


N = BICcherry(G)

color_map = []
for node in N.nodes():
    if 'color' in N.nodes[node]:
        color_map.append(N.nodes[node]['color'])
    else:
        color_map.append('gray')

for node in N.nodes():
    if node == 'rho':
        N.nodes[node]['subset'] = 0
    elif node.startswith('p_'):
        N.nodes[node]['subset'] = 1
    elif node.startswith('q_'):
        N.nodes[node]['subset'] = 2
    else:
        N.nodes[node]['subset'] = 3

pos = nx.multipartite_layout(N, subset_key='subset')

pos = {node: (-y, -x) for node, (x, y) in pos.items()}

nx.draw(N, with_labels=True, node_color=color_map, pos=pos)
#plt.show()
```

```python
# 5 leaves: a(red), b(red), c(blue), d(blue), e(green)
# e is a sicor (only green) — must be in-hub
G2 = nx.DiGraph()
G2.add_node('a', color='red')
G2.add_node('b', color='red')
G2.add_node('c', color='blue')
G2.add_node('d', color='blue')
G2.add_node('e', color='green')

# a sees both blues and green
G2.add_edge('a', 'c')
G2.add_edge('a', 'd')
G2.add_edge('a', 'e')

# b sees c (blue) and e (green), but NOT d
G2.add_edge('b', 'c')
G2.add_edge('b', 'e')

# c sees both reds and green
G2.add_edge('c', 'a')
G2.add_edge('c', 'b')
G2.add_edge('c', 'e')

# d sees a (red) and e (green), but NOT b
G2.add_edge('d', 'a')
G2.add_edge('d', 'e')

# e sees a (red) and c (blue), but NOT b and NOT d
G2.add_edge('e', 'a')
G2.add_edge('e', 'c')



color_map = []
for node in G2.nodes():
    if 'color' in G2.nodes[node]:
        color_map.append(G2.nodes[node]['color'])
    else:
        color_map.append('gray')
nx.draw(G2, with_labels=True, node_color=color_map)
plt.show()
```
