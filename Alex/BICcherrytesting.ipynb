```python
import networkx as nx
import random
import pandas as pd

G = nx.DiGraph()

for i in range(6):
    G.add_node(i, color = random.choice(["red","yellow","blue"]))
    
for i in range(6):
    for j in range(6):
        if i != j and random.random()< .3:
            G.add_edge(i,j)
            
```

```python
print(nx.to_pandas_adjacency(G))
print(len(nx.get_node_attributes(G,"color")))
print(G.edges())
```

```python
colors = nx.get_node_attributes(G,"color")
for u,v in G.edges():
    if colors[u] == colors[v]:
        print(u,v)
```
